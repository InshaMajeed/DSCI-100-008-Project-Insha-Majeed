#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# To keep only the variables needed to answer the research question
march_madness_tidy <- march_madness_data |>
    select(seed, reg_wins_pct, tourney_finish) |>

# Convert tournament finish into a categorical variable
# and relabel the categories to make them easier to understand
    mutate(
        tourney_finish = as_factor(tourney_finish),
        tourney_finish = fct_recode(
            tourney_finish,
            "Opening round elimination" = "opening_round_loss",
            "First round elimination" = "first_round_loss",
            "Second round elimination" = "second_round_loss",
            "Eliminated in final 16" = "top_16_loss",
            "Eliminated in final 8" = "top_8_loss",
            "Eliminated in final 4" = "top_4_loss",
            "Runner-up" = "top_2_loss",
            "Champion" = "champ" )) |>

# Remove observations with missing values in the variables needed for the analysis
    drop_na()

# Display the first 6 rows of the tidied dataset
head(march_madness_tidy)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
# Calculate the mean, minimum, and maximum for all numeric variables in the tidied dataset
summary_stats <- march_madness_tidy |>
    summarize(
        across(
            where(is.numeric),
            list(
                mean = ~ mean(.x, na.rm = TRUE),
                min = ~ min(.x, na.rm = TRUE),
                max = ~ max(.x, na.rm = TRUE)
            )
        )
    )

# Display the summary statistics table
summary_stats
```

```r
# Count the number of missing values in each column
missing_values <- march_madness_tidy |>
    summarize(
        across(
            everything(),
            ~ sum(is.na(.x))
        )
    )

# Display the missingness table
missing_values
```

#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

```r
# Plot the distribution of tournament seed
seed_distribution <- march_madness_tidy |>
    ggplot(aes(x = seed)) +
    geom_histogram(binwidth = 1, boundary = -0.5) +
    scale_x_continuous(breaks = 0:16) +
    labs(
        x = "Tournament seed",
        y = "Number of teams",
        title = "Distribution of Tournament Seeds"
    ) +
    theme(text = element_text(size = 12))

seed_distribution
```

```r

# Order tournament finish categories from earliest elimination to champion
finish_proportions <- march_madness_tidy |>
    mutate(
        tourney_finish = factor(
            tourney_finish,
            levels = c(
                "Opening round elimination",
                "First round elimination",
                "Second round elimination",
                "Eliminated in final 16",
                "Eliminated in final 8",
                "Eliminated in final 4",
                "Runner-up",
                "Champion"
            )
        )
    ) |>
    count(tourney_finish) |>
    mutate(proportion = n / sum(n))

# Plot the relative proportion of teams in each finish category
finish_distribution <- finish_proportions |>
    ggplot(aes(x = tourney_finish, y = proportion)) +
    geom_col() +
    labs(
        x = "Tournament finish",
        y = "Proportion of teams",
        title = "Proportion of Teams by Tournament Finish"
    ) +
    theme(
        text = element_text(size = 12),
        axis.text.x = element_text(angle = 45, hjust = 1)
    )

finish_distribution
```

### Visualization 1: Mean Tournament Seed by Tournament Finish

This visualization compares the mean tournament seed across tournament finish categories. It is relevant to the research question because tournament seed is one of the predictors being considered. Overall, teams that advance further tend to have lower mean seed numbers, suggesting that stronger seeds may be associated with greater tournament advancement. The opening-round category has a mean seed of 0 because the dataset uses 0 as a special designation for the 1983 opening-round teams rather than as a conventional tournament seed.

```r
# Order tournament finish categories by how far teams advanced
march_madness_tidy <- march_madness_tidy |>
    mutate(
        tourney_finish = factor(
            tourney_finish,
            levels = c(
                "Champion",
                "Runner-up",
                "Eliminated in final 4",
                "Eliminated in final 8",
                "Eliminated in final 16",
                "Second round elimination",
                "First round elimination",
                "Opening round elimination"
            )
        )
    )

# Calculate the mean tournament seed for each finish category
seed_finish_summary <- march_madness_tidy |>
    group_by(tourney_finish) |>
    summarize(
        mean_seed = mean(seed, na.rm = TRUE)
    )

# Plot mean tournament seed by tournament finish
seed_finish_plot <- seed_finish_summary |>
    ggplot(aes(x = mean_seed, y = tourney_finish)) +
    geom_col() +
    labs(
        x = "Mean tournament seed",
        y = "Tournament finish",
        title = "Mean Tournament Seed by Tournament Finish"
    ) +
    theme(text = element_text(size = 12))

seed_finish_plot
```

### Visualization 2: Mean Regional Win Percentage by Tournament Finish

This visualization compares the mean regional win percentage across tournament finish categories. It is relevant to the research question because regional win percentage is the second predictor being considered. Teams that advance to later stages generally have higher mean regional win percentages, suggesting that stronger regional performance may be associated with advancing further in the tournament and may therefore be useful for prediction.

```r
# Order tournament finish categories by how far teams advanced
march_madness_tidy <- march_madness_tidy |>
    mutate(
        tourney_finish = factor(
            tourney_finish,
            levels = c(
                "Champion",
                "Runner-up",
                "Eliminated in final 4",
                "Eliminated in final 8",
                "Eliminated in final 16",
                "Second round elimination",
                "First round elimination",
                "Opening round elimination"
            )
        )
    )

# Calculate the mean regional win percentage for each finish category
win_pct_finish_summary <- march_madness_tidy |>
    group_by(tourney_finish) |>
    summarize(
        mean_reg_wins_pct = mean(reg_wins_pct, na.rm = TRUE)
    )

# Plot mean regional win percentage by tournament finish
win_pct_finish_plot <- win_pct_finish_summary |>
    ggplot(aes(x = mean_reg_wins_pct, y = tourney_finish)) +
    geom_col() +
    labs(
        x = "Mean regional win percentage (%)",
        y = "Tournament finish",
        title = "Mean Regional Win Percentage by Tournament Finish"
    ) +
    theme(text = element_text(size = 12))

win_pct_finish_plot
```

I plan to use K nearest neighbors (KNN) classification to predict tournament finish (`tourney_finish`) using tournament seed (`seed`) and regional win percentage (`reg_wins_pct`). KNN is appropriate because the response variable consists of distinct tournament finish categories, making this a classification problem, while both predictors are quantitative. The exploratory analysis also suggests that both predictors are related to tournament advancement, particularly tournament seed.

I will first split the tidied data into a 75% training set and 25% testing set, stratifying by `tourney_finish` so that the relative frequency of each finish category is represented in both sets. The test set will be kept separate until the final model is selected. Within the training data, I will use 5-fold cross-validation to choose the value of \(K\) that gives the best classification performance. Because KNN calculates distances between observations, I will standardize `seed` and `reg_wins_pct` using the training data so that differences in their numerical scales do not disproportionately influence the model.

A limitation of KNN is that predictions depend strongly on the choice of \(K\), scaling of the predictors, and the availability of similar nearby observations. The tournament finish categories are also imbalanced, with some outcomes occurring much more frequently than others, which may cause the model to favour common classes. In addition, the special `seed = 0` designation used for the 1983 opening round does not represent a conventional tournament ranking and may affect distance calculations.

I will assess final performance on the test set using classification accuracy and a confusion matrix. Accuracy summarizes the overall proportion of correct predictions, while the confusion matrix will show which tournament finish categories are predicted correctly or commonly confused with one another.
